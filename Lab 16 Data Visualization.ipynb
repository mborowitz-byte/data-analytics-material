**Read the CSV File into a Pandas DataFrame**

**Create a SQLite Database and Insert the Data**

Now, let's create a new SQLite database (`survey-data.sqlite`) and insert the data from the DataFrame into a table using the sqlite3 library:


```python
import sqlite3

# Create a connection to the SQLite database
conn = sqlite3.connect('survey-data.sqlite')

# Write the dataframe to the SQLite database
df.to_sql('main', conn, if_exists='replace', index=False)


# Close the connection
conn.close()
```

**Verify the Data in the SQLite Database**
Verify that the data has been correctly inserted into the SQLite database by running a simple query:


```python
# Reconnect to the SQLite database
conn = sqlite3.connect('survey-data.sqlite')

# Run a simple query to check the data
QUERY = "SELECT * FROM main LIMIT 5"
df_check = pd.read_sql_query(QUERY, conn)

# Display the results
print(df_check)
```

For example, you can group data by a specific column, like Age, to get the count of respondents in each age group:


```python
QUERY = """
SELECT Age, COUNT(*) as count
FROM main
GROUP BY Age
ORDER BY Age
"""
pd.read_sql_query(QUERY, conn)
```

### Visualizing the Distribution of Data

**Histograms**

Plot a histogram of CompTotal (Total Compensation).


```python
## Write your code here
# Pull total compensation from the database and plot its distribution.
# Cap at the 99th percentile ensure a few extreme values do not flatten the chart.
comp = pd.read_sql_query("SELECT CompTotal FROM main", conn)['CompTotal'].dropna()
comp = comp[comp < comp.quantile(0.99)]
plt.figure(figsize=(10, 5))
plt.hist(comp, bins=100, edgecolor='black')
plt.title('Distribution of Total Compensation (CompTotal)')
plt.xlabel('CompTotal')
plt.ylabel('Frequency')
plt.show()
```

**Box Plots**

Plot a box plot of Age.


```python
## Write your code here
# Age stored as text buckets, map each bucket to a representative age.
age_map = {'Under 18 years old': 16, '18-24 years old': 21, '25-34 years old': 30,
          '35-44 years old': 40, '45-54 years old': 50, '55-64 years old': 60, 
          '65 years or older': 70, 'Prefer not to say': None}
ages = pd.read_sql_query("SELECT Age FROM main", conn)['Age'].map(age_map).dropna()
plt.figure(figsize=(6, 5))
plt.boxplot(ages)
plt.title('Box Plot of Respondent Age')
plt.ylabel('Age (years)')
plt.show()

```

### Visualizing Relationships in Data

**Scatter Plots**

Create a scatter plot of Age and WorkExp.


**Bubble Plots**

Create a bubble plot of `TimeSearching` and `Frustration` using the Age column as the bubble size.


```python
## Write your code here
# TimeSearching is ordinatl (map to minutes); use the number of reported
# frustrations as a numeric measure and Age as the bubble size.
age_map = {'Under 18 years old': 16, '18-24 years old': 21, '25-34 years old': 30,
           '35-44 years old': 40, '45-54 years old': 50, '55-64 years old': 60,
           '65 years or older': 70, 'Prefer not to say': None}
time_map = {'Less than 15 minutes a day': 7.5, '15-30 minutes a day': 22.5,
            '30-60 minutes a day': 45, '60-120 minutes a day': 90,
            'Over 120 minutes a day': 150}

data = pd.read_sql_query("SELECT TimeSearching, Frustration, Age FROM main", conn)
data['TimeSearching_min'] = data['TimeSearching'].map(time_map)
data['Frustration_count'] = data['Frustration'].str.split(';').str.len()
data['Age_numeric'] = data['Age'].map(age_map)
data = data.dropna(subset=['TimeSearching_min', 'Frustration_count', 'Age_numeric'])

plt.figure(figsize=(9, 6))
plt.scatter(data['TimeSearching_min'], data['Frustration_count'], s=data['Age_numeric'], alpha=0.2)
plt.title('Time Searching vs Number of Frustrations (bubble size = Age)')
plt.xlabel('Time Searching (minutes a day)')
plt.ylabel('Number of Frustraions reported')
plt.show()

```

### Visualizing Composition of Data

**Pie Charts**

Create a pie chart of the top 5 databases(`DatabaseWantToWorkWith`) that respondents wish to learn next year.


```python
## Write your code here
# DatabaseWantToWorkWith is multi-select; split, explode, take top 5.
dbs = pd.read_sql_query("SELECT DatabaseWantToWorkWith FROM main", conn)['DatabaseWantToWorkWith']
top5 = dbs.dropna().str.split(';').explode().str.strip().value_counts().head(5)

plt.figure(figsize=(7, 7))
plt.pie(top5, labels=top5.index, autopct='%1.1f%%', startangle=90)
plt.title('Top 5 Databases Respondents Wnat to Work With')
plt.show()
```

**Stacked Charts** 

Create a stacked bar chart of median `TimeSearching` and `TimeAnswering` for the age group 30 to 35.


```python
## Write your code here
# Map both time columns to minutes, then stack the medians for the 30-35 age group.
age_map = {'Under 18 years old': 16, '18-24 years old': 21, '25-34 years old': 30,
           '35-44 years old': 40, '45-54 years old': 50, '55-64 years old': 60,
           '65 years or older': 70, 'Prefer not to say': None}
time_map = {'Less than 15 minutes a day': 7.5, '15-30 minutes a day': 22.5,
            '30-60 minutes a day': 45, '60-120 minutes a day': 90,
            'Over 120 minutes a day': 150}

data = pd.read_sql_query("SELECT Age, TimeSearching, TimeAnswering FROM main", conn)
data['Age_numeric'] = data['Age'].map(age_map)
data['TimeSearching_min'] = data['TimeSearching'].map(time_map)
data['TimeAnswering_min'] = data['TimeAnswering'].map(time_map)
group = data[(data['Age_numeric'] >= 30) & (data['Age_numeric'] <=35)]
medians = group[['TimeSearching_min', 'TimeAnswering_min']].median()

plt.figure(figsize=(5, 6))
plt.bar('Ages 30-35', medians['TimeSearching_min'], label='Time Searching')
plt.bar('Ages 30-35', medians['TimeAnswering_min'], bottom=medians['TimeSearching_min'], label='Time Answering')
plt.ylabel('Minutes per day (median)')
plt.legend()
plt.show()
```

### Visualizing Comparison of Data

**Line Chart**

Plot the median `CompTotal` for all ages from 45 to 60.


```python
## Write your code here
# Group by age midpoint and track the median compensation across the 45-60 range
age_map = {'Under 18 years old': 16, '18-24 years old': 20, '25-34 years old': 30,
           '35-44 years old': 40, '45-54 years old': 50, '55-64 years old': 60,
           '65 years or older': 70, 'Prefer not to say': None}

data = pd.read_sql_query("SELECT Age, CompTotal FROM main", conn)
data['Age_numeric'] = data['Age'].map(age_map)
subset = data[(data['Age_numeric'] >= 45) & (data['Age_numeric'] <=60)]
median_comp = subset.groupby('Age_numeric')['CompTotal'].median()

plt.figure(figsize=(8,5))
plt.plot(median_comp.index, median_comp.values, marker='o')
plt.title('Median Total Compensation for Ages 45-60')
plt.xlabel('Age')
plt.ylabel('Median CompTotal')
plt.show()
```

**Bar Chart**

Create a horizontal bar chart using the `MainBranch` column.


```python
## Write your code here
# Count respondents in each MainBranch category and show as a horizontal bar chart.
counts = pd.read_sql_query("SELECT MainBranch FROM main", conn)['MainBranch'].value_counts()

plt.figure(figsize=(9,5))
plt.barh(counts.index, counts.values)
plt.title('Distribution of MainBranch')
plt.xlabel('Number of respondents')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```
